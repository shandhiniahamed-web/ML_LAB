```python
# Book Recommendation using Collaborative Filtering

import pandas as pd

# Sample Book Recommendation Dataset
data = {
    "User": [
        "User1", "User1", "User1",
        "User2", "User2", "User2", "User2",
        "User3", "User3", "User3",
        "User4", "User4", "User4",
        "User5", "User5", "User5"
    ],

    "Book": [
        "Harry Potter", "The Alchemist", "Atomic Habits",
        "Harry Potter", "The Alchemist", "Atomic Habits", "Rich Dad Poor Dad",
        "Harry Potter", "Atomic Habits", "Rich Dad Poor Dad",
        "The Alchemist", "Atomic Habits", "Rich Dad Poor Dad",
        "Harry Potter", "The Alchemist", "Rich Dad Poor Dad"
    ],

    "Rating": [
        5, 4, 5,
        5, 4, 4, 5,
        4, 5, 4,
        3, 5, 5,
        5, 3, 5
    ]
}

df = pd.DataFrame(data)

print("Book Recommendation Dataset:")
print(df)

# Create User-Book Rating Matrix
rating_matrix = df.pivot_table(
    index="User",
    columns="Book",
    values="Rating"
)

print("\nUser-Book Rating Matrix:")
print(rating_matrix.fillna(0))

# Target user
target_user = "User1"

# Find books rated by target user
target_ratings = rating_matrix.loc[target_user]

# Calculate average rating for each book
average_ratings = rating_matrix.mean()

# Books already read/rated by target user
rated_books = target_ratings.dropna().index

# Remove already rated books
recommendations = average_ratings.drop(rated_books)

# Sort books based on average rating
recommendations = recommendations.sort_values(ascending=False)

print("\nTarget User:", target_user)

print("\nRecommended Books:")

if len(recommendations) > 0:
    for book, rating in recommendations.items():
        print(book, "-> Average Rating:", round(rating, 2))
else:
    print("No new books available for recommendation.")

# Final recommendation
if len(recommendations) > 0:
    best_book = recommendations.index[0]
    print("\nRecommended Book:", best_book)
```
